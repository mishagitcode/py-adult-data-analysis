# Task 0
Read the dataset from csv file & perform data cleaning - remove all rows, which contains `?` in some columns.
Also check for data correctness (salary & salary $K).

```python
data = pd.read_csv('../data/adult.csv')

data_clean = data.replace("?", pd.NA).dropna()

if "salary" in data_clean.columns and "salary$K" in data_clean.columns:
    data_clean["salary$K_calc"] = data_clean["salary"] / 1000

    incorrect_rows = data_clean[
        data_clean["salary$K_calc"] != data_clean["salary$K"]
    ]

    print("Number of incorrect salary rows:", len(incorrect_rows))
else:
    print("Salary columns not found in dataset.")
  
data_clean = data_clean.drop(columns=["salary$K_calc"], errors="ignore")

data.info()

data.head()
```

# Task 4
Get the mean and standard deviation of the age for people who earn > 50K per year. After this, get it for those who earn <= 50K.

# Task 5
Check, if there are some people without higher education (education: Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters, Doctorate), but with > 50K salary

```python
higher_edu = [
    "Bachelors",
    "Prof-school",
    "Assoc-acdm",
    "Assoc-voc",
    "Masters",
    "Doctorate"
]

# Filter rows: NOT in higher_edu AND salary > 50K
result = data[
    (~data["education"].isin(higher_edu)) &
    (data["salary"] == ">50K")
]

result
```

# Task 7
Compare the married and non-married men salaries. Who earns more? (>50K or <=50K)
Married men are those, whom `marital-status` starts with "Married". Others are not.

```python
men = data[data["sex"] == "Male"]
married_mask = men["marital-status"].str.startswith("Married")

married_men = men[married_mask]
non_married_men = men[~married_mask]

married_salary = married_men["salary"].value_counts()
non_married_salary = non_married_men["salary"].value_counts()

married_salary, non_married_salary
```

# Task 8
Get the max hours per week some person works. How many people works the same amount of hours per week?

```python
max_hours = data["hours-per-week"].max()
count_max_hours = (data["hours-per-week"] == max_hours).sum()

max_hours, count_max_hours
```
